## 2️⃣ Import Sysmic Core

```python
# Core framework
from sysmic.system import Sysmic
from sysmic.infrastructure import SystemConfiguration, CertificationLevel
from sysmic.data import SeismicDataAcquisition
from sysmic.vis import StyleManager

# Utilities
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
StyleManager.set_premium_style()

print("✅ Sysmic Framework imported successfully!")
```

## 3️⃣ Configuration & Data Fetch

Select region and certification level.

```python
# ========== CONFIGURATION ==========

REGION = "San Andreas Fault"
LEVEL = "LEVEL_2"  # Options: LEVEL_0 (Basic) to LEVEL_4 (Full Audit)

# Initialize System
config = SystemConfiguration(
    certification_level=CertificationLevel[LEVEL]
)
sysmic = Sysmic(config)

print(f"⚙️ System Initialized: {sysmic.config.system_id}")
print(f"   Level: {LEVEL}")
```

```python
# Fetch Data
print("🌍 Fetching data from USGS...")
data_acq = SeismicDataAcquisition()

# San Andreas Preset bounds (approx)
bounds = (32.0, 38.0, -122.0, -115.0)

catalog_raw = data_acq.fetch_usgs_catalog(
    bounds=bounds,
    min_magnitude=2.5,
    start_date="2010-01-01",
    end_date="2025-12-31"
)

print(f"✅ Loaded {len(catalog_raw)} events")
catalog_raw.head()
```

## 4️⃣ Run Full Analysis

Execute the Sysmic pipeline.

```python
print("🚀 Starting Sysmic Analysis...")

# Define analysis profile based on level
types = ['fractal_dimension', 'temporal', 'multifractal', 'validation']

# EXECUTE
result = sysmic.analyze_catalog(
    catalog_raw, 
    analysis_types=types
)

if result.success:
    print("✅ Analysis Complete!")
    print(f"   Computation Time: {result.computation_time:.2f}s")
    print(f"   Analysis ID: {result.analysis_id}")
else:
    print("❌ Analysis Failed")
    print(result.errors)
```

## 5️⃣ Results & Visualization

```python
if result.success:
    # 1. Fractal Dimension
    if result.fractal_dimension:
        cons = result.fractal_dimension.get('consensus', {})
        d_val = cons.get('value', 0)
        unc = cons.get('uncertainty', 0)
        print(f"\n📊 Fractal Dimension D = {d_val:.3f} ± {unc:.3f}")
        
        # Simple interpretation
        if d_val < 2.0: print("   → Planar (Fault-constrained)")
        elif d_val > 2.5: print("   → Volumetric (Distributed)")
        else: print("   → Intermediate (Complex Fault Network)")

    # 2. Multifractal
    if result.multifractal_spectrum:
        spec = result.multifractal_spectrum.get('spectrum_analysis', {})
        width = spec.get('spectrum_width', 0)
        print(f"\n🔬 Multifractal Width Δα = {width:.3f}")

    # 3. Report
    print("\n📝 Generating ISO-9001 Report...")
    report = result.generate_report()
    print(report)
```

## 6️⃣ Interactive Visualization

Visualize the correlation integral.

```python
if result.fractal_dimension and 'diagnostics' in result.fractal_dimension:
    diag = result.fractal_dimension['diagnostics']
    
    plt.figure(figsize=(10, 6))
    plt.loglog(diag['r'], diag['C_r'], 'o-', linewidth=2, label='Correlation Integral')
    plt.xlabel('Distance r')
    plt.ylabel('C(r)')
    plt.title(f'Fractal Dimension D2 = {d_val:.3f}')
    plt.grid(True, which="both", ls="-", alpha=0.2)
    plt.legend()
    plt.show()
```
